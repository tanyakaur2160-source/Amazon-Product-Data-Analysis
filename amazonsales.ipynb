```python
df['discounted_price']=(
    df['discounted_price']
    .astype(str)
    .str.replace('₹','', regex=False)
    .str.replace(',','', regex=False)
    .str.strip()
)

df['actual_price']=(
    df['actual_price']
    .astype(str)
    .str.replace('₹','', regex=False)
    .str.replace(',','', regex=False)
    .str.strip()
)
df['discounted_price']=pd.to_numeric(
    df['discounted_price'],
    errors='coerce'
)
df['actual_price']=pd.to_numeric(
    df['actual_price'],
    errors='coerce'
)
```

```python
df[['actual_price','discounted_price']].head()
```

```python
df['discount_percentage']=(
    df['discount_percentage']
    .astype(str)
    .str.replace('%','', regex=False)
)
df['discount_percentage']=pd.to_numeric(
    df['discount_percentage'],
    errors='coerce'
)
```

```python
df['discount_percentage'].describe()
```

```python
df['rating']=pd.to_numeric(
    df['rating'],
    errors='coerce'
)
```

```python
df['rating'].value_counts().sort_index()
```

```python
df['rating_cunt']=(
    df['rating_count']
    .astype(str)
    .str.replace(',','', regex=False)

)
df['rating_count']=pd.to_numeric(
    df['rating_count'],
    errors='coerce'
)
```

```python
df['rating_count'].describe()
```

```python
df['savings']=(
    df['actual_price']-df['discounted_price']
)
```

```python
df['calculated_discount']=(
    df['savings']/df['actual_price']*100

)
```

```python
df['value_score'] = (
    df['rating'] * np.log1p(df['rating_count'])
)
df.nlargest(
    20,
    'value_score'
)[
    [
        'product_name',
        'rating',
        'rating_count',
        'discount_percentage',
        'discounted_price',
        'value_score'
    ]
]
```

```python
bins = [0, 20, 40, 60, 80, 100]

labels = [
    '0-20%',
    '21-40%',
    '41-60%',
    '61-80%',
    '81-100%'
]

df['discount_bucket'] = pd.cut(
    df['discount_percentage'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
discount_analysis = (
    df.groupby('discount_bucket', observed=True)
    .agg(
        products=('product_id', 'count'),
        avg_rating=('rating', 'mean'),
        avg_price=('discounted_price', 'mean'),
        total_reviews=('rating_count', 'sum')
    )
)

discount_analysis.round(2)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=discount_analysis.reset_index(),
    x='discount_bucket',
    y='avg_rating'
)

plt.title('Average Rating by Discount Range')
plt.xlabel('Discount Range')
plt.ylabel('Average Rating')

plt.show()
```
